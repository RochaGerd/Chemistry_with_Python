## <font color='magenta'>**1.5 Dois exemplos rápidos de físico-química (opcional)**</font>

Dois gráficos clássicos, só para mostrar quanta física cabe em vinte linhas de NumPy. O primeiro é a distribuição de velocidades de Maxwell-Boltzmann para o CO₂ em duas temperaturas; o segundo, as funções de onda da partícula na caixa unidimensional.

Fonte dos problemas: P. Atkins, L. Jones, L. Laverman. *Princípios de Química: questionando a vida moderna e o meio ambiente*, 7ª ed., Bookman, 2018 (tópicos 3D.3 e 1C).

````python
#@title <font color='magenta'> 1.5.1 — Distribuição de velocidades de Maxwell-Boltzmann
# Exemplo Distribuição de Velocidades moleculares em um gás ideal
# A distribuição das velocidades das moléculas de um gás ideal pode ser calculada com base no tratamento dado por James Clerk Maxwell juntamente com Ludwig Boltzmann.
# Considere 1 mol de gás carbônico (CO2) inserido em um recipiente de volume fixo de 1L.
# Obtenha as curvas de distribuição de velocidade molecular para essa amostra nas temperaturas de 300K e 800K.
# Fonte para consulta: Tópico 3D.3 do livro Peter Atkins, Loretta Jones e Lerroy Laverman, Princípios de Química: questionando a vida moderna e o meio ambiente, 7.ed., Porto Alegre, Bookman, 2018.
import numpy as np
import matplotlib.pyplot as plt

M_CO2 = 44.0095  # Em g/mol
R = 8.3145 # J/(K*mol)
T1 = 300  # K
T2 = 800  # K
ct1 = ((M_CO2/1000.0)/(2.0*R*T1))
ct2 = ((M_CO2/1000.0)/(2.0*R*T2))

npt = 100000
vmin, vmax = 0.01, 2000.0
v = np.linspace(vmin, vmax, npt)

fr1 = 4.0*np.pi*((ct1/np.pi)**(1.5))*np.square(v)*np.exp(-ct1*np.square(v))
fr2 = 4.0*np.pi*((ct2/np.pi)**(1.5))*np.square(v)*np.exp(-ct2*np.square(v))

plt.plot(v,fr1, label="T = 300K", lw = 2.5)
plt.plot(v,fr2, label="T = 800K", lw = 2.5)
plt.title("Gráfico de distribuições de velocidades moleculares ", fontsize=16)
plt.xlabel("Velocidade molecular (m/s)",fontsize=14)
plt.ylabel("Fração, f(v) (s/m)",fontsize=14)
plt.legend()
plt.tight_layout()
plt.savefig('molecular_speed.png',dpi=400, transparent=True)
plt.show()
````

````python
#@title <font color='magenta'> 1.5.2 — Partícula na caixa unidimensional
# Expemplo 5.5
# Um elétron (me = 9,1 x 10 ^-31 kg) está confinado em uma caixa unidimensional de tamanho 50 nM.
# Produza os gráficos da função de onda do estado fundamental e dos três primeiros estados excitados em uma mesma figura.
# Fonte para consulta: Tópico 1C do livro Peter Atkins, Loretta Jones e Lerroy Laverman, Princípios de Química: questionando a vida moderna e o meio ambiente, 7.ed., Porto Alegre, Bookman, 2018.
import numpy as np
import matplotlib.pyplot as plt

L = 50.0e-9                               # tamanho da caixa (m)
a0 = 5.2917721067e-11                             # raio de Bohr
L_a0 = L / a0             # tamanho da caixa (unidades atômicas)

me = 9.1e-31                            # massa do elétron em kg
xmin, xmax = 0, L_a0                           # bordas da caixa
npt = 1000                            # número de pontos da função
x = np.linspace(xmin, xmax, npt)                  # grid de pontos

A = np.sqrt(2/L_a0)
psi_n1 = A * np.sin(1 * np.pi * x / L_a0)    # psi(x) para n = 1
psi_n2 = A * np.sin(2 * np.pi * x / L_a0)    # psi(x) para n = 2
psi_n3 = A * np.sin(3 * np.pi * x / L_a0)    # psi(x) para n = 3
psi_n4 = A * np.sin(4 * np.pi * x / L_a0)    # psi(x) para n = 4
psi_n5 = A * np.sin(5 * np.pi * x / L_a0)    # psi(x) para n = 5

fig = plt.figure()

plt.subplot(3, 1, 1)
plt.plot(x, psi_n5,'-k', lw = 2.0)
plt.xlabel('x (em Bohr)')
plt.ylabel(r'$\Psi_{5}(x)$')
plt.plot((0,xmax),(0,0),'--')       # linha horizontal tracejada

plt.subplot(3, 2, 3)
plt.plot(x, psi_n4,'-b', lw = 2.0)
plt.xlabel('x (em Bohr)')
plt.ylabel(r'$\Psi_{4}(x)$')
plt.plot((0,xmax),(0,0),'--')       # linha horizontal tracejada

plt.subplot(3, 2, 4)
plt.plot(x, psi_n3, '-r', lw = 2.0)
plt.xlabel('x (em Bohr)')
plt.ylabel(r'$\Psi_{3}(x)$')
plt.plot((0,xmax),(0,0),'--')       # linha horizontal tracejada

plt.subplot(3, 2, 5)
plt.plot(x, psi_n2, '-g', lw = 2.0)
plt.xlabel('x (em Bohr)')
plt.ylabel(r'$\Psi_{2}(x)$')
plt.plot((0,xmax),(0,0),'--')       # linha horizontal tracejada

plt.subplot(3, 2, 6)
plt.plot(x, psi_n1, '-y', lw = 2.0)
plt.xlabel('x (em Bohr)')
plt.ylabel(r'$\Psi_{1}(x)$')

plt.tight_layout()

plt.show()
````

````python
#@title <font color='magenta'> 1.5.3 — Partícula na caixa bidimensional
import numpy as np
import matplotlib.pyplot as plt

L, nx, ny = 1, 2, 2
x = np.linspace(0, L, 20)
y = np.linspace(0, L, 20)
X, Y = np.meshgrid(x,y)

Psi = (2/L) * np.sin(nx*np.pi*X/L) * np.sin(ny*np.pi*Y/L)

fig = plt.figure(figsize=(10,6))
plt.contourf(X, Y, Psi, 12, cmap='jet', alpha=0.8)
contorno = plt.contour(X, Y, Psi, 12, colors='black')
plt.clabel(contorno)

plt.title(r'Gráfico 2D de $\Psi_{2,2}(X,Y)$', fontsize=16)
plt.ylabel('Y', fontsize = 14)
plt.xlabel('X', fontsize = 14)
plt.savefig('2D_part_in_box_600.png', dpi=600, transparent=True)
````

# <font color='green'> **3. Fundamentos: como o computador representa uma molécula**

## <font color='green'>**3.1 O que é modelagem molecular**</font>

A IUPAC define assim:

> <font color='red'>Modelagem molecular é a investigação das estruturas e das propriedades moleculares pelo uso da química computacional e de técnicas de visualização gráfica, visando fornecer uma representação tridimensional sob um dado conjunto de circunstâncias.</font>

[Verbete no GoldBook da IUPAC](https://goldbook.iupac.org/terms/view/MT06971)

Reparem em "sob um dado conjunto de circunstâncias". Uma molécula não tem *uma* estrutura. Ela tem a estrutura compatível com a temperatura, o solvente e o alvo com que está interagindo. Boa parte dos erros que vejo em trabalhos de modelagem vem de esquecer essa cláusula.

O trabalho tem três movimentos: construir o modelo, calcular propriedades e fenômenos a partir dele, e usar o resultado para decidir o que sintetizar ou produzir. É uma área de fronteira, que envolve pesquisadore(a)s de física, química, matemática, computação, estatística, farmácia, biologia, engenharias e ciências da saúde.

Quando o experimento acontece dentro do computador, num sistema modelo, chamamos de *in silico*, por analogia com *in vitro* e *in vivo*, e em homenagem ao silício dos chips.

## <font color='green'>**3.2 Do nome à superfície: os níveis de representação**</font>

Antes de calcular qualquer coisa é preciso decidir como a molécula entra no computador. Existe uma escada de representações, cada degrau carregando mais informação e custando mais caro computacionalmente:

```
      1D              1D texto            2D                3D                3D

 │     e     │───►│    InChI     │──►│             │──►│             │─►│             │
 │  FÓRMULA  │    │              │   │  átomos +   │   │   x, y, z   │  │  densidade  │
 │           │    │ CC(=O)Oc1ccc │   │  ligações   │   │             │  │ eletrônica  │
 │  C9H8O4   │    │ cc1C(=O)O    │   │             │   │ distâncias  │  │  potencial  │
 │           │    │              │   │   grupos    │   │  ângulos    │  │eletrostático│
 │           │    │              │   │ funcionais  │   │  diedros    │  │  área SASA  │
 └───────────┘    └──────────────┘   └─────────────┘   └─────────────┘  └─────────────┘

 └────────────────────────────────────────────────────────────────────────────────────┘
     mais informação estrutural  ──────────────────────────────►  mais custo de cálculo
```

- **1D — nome ou fórmula.** Identifica o composto e permite classificá-lo. Não diz nada sobre estrutura.
- **1D em texto — SMILES, InChI.** Uma linha de caracteres que codifica a conectividade inteira. É o formato de troca entre bancos de dados e programas.
- **2D — fórmula plana ou de linhas.** Explicita ligações e grupos funcionais. É o grafo molecular: átomos são vértices, ligações são arestas.
- **3D — coordenadas atômicas.** Distâncias, ângulos, diedros, conformação. É aqui que a molécula ganha forma e passa a poder encaixar (ou não) num sítio ativo.
- **3D — superfícies.** Mapas de densidade eletrônica, potencial eletrostático, hidrofobicidade, área acessível ao solvente. É o que o alvo biológico "enxerga".

A escolha da representação não é detalhe técnico: ela define o custo do cálculo e limita o que você consegue concluir. Um modelo QSAR construído sobre descritores 2D nunca vai distinguir dois enantiômeros. Nem sempre isso importa. Quando importa, importa muito, e a talidomida está aí para lembrar.

Cada representação deu origem a formatos de arquivo diferentes, e cada programa tem seu preferido. Para converter entre eles existe o [OpenBabel](https://openbabel.org/index.html), que resolve praticamente qualquer par de formatos. Há também uma [versão online](https://www.cheminfo.org/Chemistry/Cheminformatics/FormatConverter/index.html) para conversões rápidas.

## <font color='green'>**3.3 O grafo molecular e as notações lineares**</font>

A representação canônica em quimioinformática é o **grafo** $G = (V, E)$: átomos como vértices rotulados por elemento, carga formal, hibridação e contagem de hidrogênios; ligações como arestas rotuladas por ordem e aromaticidade.

Essa escolha carrega uma decisão que vale explicitar. O grafo é uma abstração **topológica**, sem coordenadas: captura conectividade e perde geometria. É por isso que quimioinformática e modelagem molecular são complementares. O grafo é barato e suficiente para buscar e classificar, e insuficiente para descrever reconhecimento molecular.

**SMILES** (Weininger, 1988) serializa o grafo numa cadeia de caracteres, percorrendo a estrutura em profundidade, com parênteses para ramificações e números para fechamento de anéis. A importância prática é enorme: uma linha de texto cabe numa célula de planilha, num índice de banco de dados, numa URL.

Duas extensões merecem menção:

- **SMARTS** estende o SMILES para consultas, com curingas, lógica booleana e restrições de grau, carga e vizinhança. É a linguagem em que se escrevem alertas estruturais, regras de tipagem e filtros.
- **SMIRKS** e as reações em SMARTS descrevem transformações, mapeando átomos do reagente no produto. É a base da enumeração de bibliotecas combinatórias e das ferramentas de retrossíntese.

**InChI** (Heller e colaboradores, 2015) segue outra filosofia. Não é legível nem projetado para ser escrito à mão: é um identificador gerado por algoritmo determinístico, em camadas (fórmula, conectividade, hidrogênios, carga, estereoquímica, isótopos). O **InChIKey**, seu resumo criptográfico de comprimento fixo, permite verificar identidade química entre bases de dados por simples comparação de chaves.

As regras básicas de SMILES cabem em seis linhas:

- Átomos são símbolos químicos: `C`, `N`, `O`, `S`.
- Hidrogênios ligados a carbono ficam implícitos.
- Ligações: simples é implícita, dupla é `=`, tripla é `#`.
- Ramificações vão entre parênteses.
- Anéis abrem e fecham com o mesmo número: `C1CCCCC1` é ciclo-hexano.
- Aromáticos vão em minúscula: `c1ccccc1` é benzeno.

| Composto | SMILES |
|---|---|
| Etileno | `C=C` |
| Etanol | `CCO` |
| Benzeno | `c1ccccc1` |
| Aspirina | `CC(=O)Oc1ccccc1C(=O)O` |
| Cafeína | `CN1C=NC2=C1C(=O)N(C)C(=O)N2C` |

Onde SMILES aparece no dia a dia: armazenamento em bancos como [PubChem](https://pubchem.ncbi.nlm.nih.gov/) e ChEMBL, busca por subestrutura, entrada de programas de predição, e formato de troca em praticamente toda API de química.

## <font color='green'>**3.4 Identidade química: um problema mais difícil do que parece**</font>

Uma mesma molécula admite muitas representações. O benzeno pode ser escrito como `c1ccccc1`, `C1=CC=CC=C1`, ou começando por qualquer átomo. `OCC`, `CCO` e `C(C)O` são todas etanol. Para decidir se dois registros são a mesma substância é preciso **canonicalizar**: escolher, entre todas as serializações possíveis, uma única de forma determinística.

O procedimento remonta ao algoritmo de Morgan (1965), que atribui a cada átomo um invariante inicial e o refina iterativamente pela soma dos invariantes dos vizinhos, até estabilizar. Empates residuais são resolvidos por regras de desempate. Implementações modernas usam variantes mais sofisticadas, e é isso que o RDKit devolve quando você pede a SMILES canônica.

Três situações continuam sendo fonte permanente de erro:

**Tautomeria.** Formas ceto e enol, ou os tautômeros de um anel imidazólico, são a mesma substância em solução e grafos diferentes. Sem canonicalização de tautômero, o mesmo composto aparece como duas entradas distintas, e modelos treinados sobre isso aprendem ruído.

**Protonação.** O estado de ionização depende do pH e não está no arquivo. Um carboxilato e o ácido correspondente são grafos diferentes para a mesma espécie em equilíbrio.

**Estereoquímica.** Centros não especificados, misturas racêmicas registradas como um único composto, atividade medida sobre a mistura. Enantiômeros podem diferir em ordens de grandeza de potência, e tratá-los como equivalentes contamina qualquer análise. A talidomida está aí para lembrar.

Nenhum desses problemas tem solução automática perfeita. O que existe são convenções de padronização, e a exigência mínima é aplicá-las de forma consistente a todo o conjunto. Se você comparar duas bases por igualdade de string sem canonizar antes, vai concluir que compostos idênticos são diferentes. Pode acontecer de um projeto travar semanas por causa disso.

## <font color='green'>**3.6 Superfície de energia potencial**</font>

Este é o conceito que sustenta tudo que vem depois: docking, dinâmica molecular, otimização de geometria, cálculo de barreira de reação. Vale gastar cinco minutos aqui.

A superfície de energia potencial (PES, *potential energy surface*) é a energia do sistema em função das posições dos núcleos:

$$E = E(\mathbf{R}_1, \mathbf{R}_2, \ldots, \mathbf{R}_N)$$

Ela nasce da aproximação de Born-Oppenheimer: como os núcleos são milhares de vezes mais pesados que os elétrons, uma forma simplificada de visualizar essa aproximação é pensar que dá para congelar os núcleos, resolver a equação de Schrödinger eletrônica para aquela geometria fixa,

$$\hat{H}\,\Psi = E\,\Psi$$

e repetir o procedimento para cada configuração nuclear. O conjunto dessas energias, ponto a ponto, é a superfície. A energia total soma a parte eletrônica e a repulsão entre núcleos:

$$E_{\text{potencial}} = E_{\text{eletrônica}} + E_{\text{nuclear}}$$

Para uma molécula com $N$ átomos, a PES tem $3N - 6$ dimensões. Ninguém desenha isso. O que se faz é olhar cortes: uma coordenada de reação, um ângulo diedro, uma distância de ligação.

**Os pontos que interessam**

- **Mínimo global** — a estrutura mais estável do sistema.
- **Mínimos locais** — estruturas estáveis, mas de energia maior. Confôrmeros, tautômeros, isômeros.
- **Pontos de sela** — estados de transição. Máximo ao longo da coordenada de reação e mínimo em todas as outras direções.

A energia de ativação é simplesmente a diferença entre o ponto de sela e o reagente:

$$E_{\text{ativação}} = E_{\text{TS}} - E_{\text{reagente}}$$

Otimizar uma geometria é descer a superfície até um mínimo. Fazer dinâmica molecular é percorrê-la com energia cinética. Fazer docking é procurar o mínimo da superfície de interação entre ligante e proteína. São três perguntas diferentes sobre o mesmo objeto.

A célula abaixo desenha um corte unidimensional com todos esses elementos.

````python
#@title <font color='green'> 3.6.1 — Um corte da superfície de energia potencial
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 1200)
V = (
    -3.0 * np.exp(-((x - 2.0) ** 2) / 0.8)
    - 2.2 * np.exp(-((x - 7.2) ** 2) / 1.2)
    + 0.6 * np.exp(-((x - 4.6) ** 2) / 1.5)
)

# Localiza os pontos críticos numericamente
i_glob = int(np.argmin(np.where(x < 4.0, V, 0)))  # mínimo global
i_loc = int(np.argmin(np.where(x > 5.5, V, 0)))  # mínimo local
faixa = (x > 3.5) & (x < 6.0)
i_ts = int(np.arange(len(x))[faixa][np.argmax(V[faixa])])  # ponto de sela

fig, ax = plt.subplots(figsize=(9.5, 5.5))
ax.plot(x, V, lw=2.6, color="#2c3e50")

# Expansão dos limites do eixo Y para dar margem vertical
ax.set_ylim(-3.8, 1.5)

# Configuração dos pontos críticos
# Nota: 'mínimo local' foi posicionado acima do ponto para não colidir com a nova seta
pontos = [
    (i_glob, "Mínimo Global", "#1a9850", (0, -18), "top"),
    (i_loc, "Mínimo Local", "#4575b4", (0, -70), "bottom"),
    (
        i_ts,
        "Ponto de Sela\n(estado de transição)",
        "#d73027",
        (0, 18),
        "bottom",
    ),
]

for i, rot, cor, offset, va in pontos:
  ax.plot(
      x[i],
      V[i],
      "o",
      ms=11,
      color=cor,
      zorder=5,
      markeredgecolor="k",
      markeredgewidth=0.6,
  )
  ax.annotate(
      rot,
      (x[i], V[i]),
      textcoords="offset points",
      xytext=offset,
      ha="center",
      va=va,
      fontsize=10,
      color=cor,
      fontweight="bold",
  )

# 1. Energia de ativação: do mínimo global até o ponto de sela
ax.annotate(
    "",
    xy=(x[i_ts], V[i_ts]),
    xytext=(x[i_ts], V[i_glob]),
    arrowprops=dict(arrowstyle="<->", color="#d73027", lw=1.8),
)
ax.text(
    x[i_ts] + 0.18,
    (V[i_ts] + V[i_glob]) / 2,
    r"$E_{\mathrm{ativação}}$",
    color="#d73027",
    fontsize=11,
    va="center",
)

# 2. Delta E conformacional: do mínimo local até o patamar do mínimo global
ax.annotate(
    "",
    xy=(x[i_loc], V[i_loc]),
    xytext=(x[i_loc], V[i_glob]),
    arrowprops=dict(arrowstyle="<->", color="#4575b4", lw=1.8),
)
ax.text(
    x[i_loc] + 0.18,
    (V[i_loc] + V[i_glob]) / 2,
    r"$\Delta E_{\mathrm{conformacional}}$",
    color="#4575b4",
    fontsize=11,
    va="center",
)

# Linhas auxiliares pontilhadas
#ax.hlines(V[i_loc], x[i_ts], x[i_loc], color="#4575b4", ls=":", lw=1.2)
ax.hlines(
    V[i_glob], x[i_glob], x[i_loc], color="#1a9850", ls=":", lw=1.2
)  # Estendida até x[i_loc]

# Ajuste nos rótulos e títulos
ax.set_xlabel(
    "Coordenada de Reação (diedro, ângulo ou distância de ligação)",
    fontsize=11,
    labelpad=12,
)
ax.set_ylabel("Energia Potencial", fontsize=11)
ax.set_title(
    "Corte unidimensional de uma superfície de energia potencial",
    fontsize=13,
    pad=20,
)

ax.set_yticklabels([])
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print("Otimizar geometria  = descer até um mínimo (local ou global).")
print("Dinâmica molecular  = passear pela superfície com energia cinética.")
print(
    "Docking              = achar o mínimo da superfície de interação"
    " ligante-proteína."
)
````

## <font color='green'>**3.7 Campos de força**</font>

Resolver a equação de Schrödinger para cada geometria é caro. Para uma proteína solvatada, com dezenas de milhares de átomos e milhões de passos de simulação, é inviável.

A saída é substituir a mecânica quântica por uma expressão clássica ajustada: o campo de força. Ele escreve a energia potencial como uma soma de termos com significado físico direto:

$$E_{\text{total}} = E_{\text{lig}} + E_{\text{ang}} + E_{\text{tor}} + E_{\text{não-lig}}$$

**Estiramento de ligação** — uma mola harmônica, lei de Hooke:

$$E_{\text{lig}} = \sum_{\text{ligações}} \frac{1}{2} k_b (r - r_0)^2$$

**Deformação angular** — outra mola, agora no ângulo entre três átomos:

$$E_{\text{ang}} = \sum_{\text{ângulos}} \frac{1}{2} k_\theta (\theta - \theta_0)^2$$

**Torção** — periódica por natureza, escrita como série de Fourier no ângulo diedro $\phi$:

$$E_{\text{tor}} = \sum_{\text{torções}} \sum_n \frac{1}{2} V_n \left[ 1 + \cos(n\phi - \gamma) \right]$$

**Interações não ligadas** — van der Waals pelo potencial de Lennard-Jones, mais a eletrostática de Coulomb entre cargas parciais:

$$E_{\text{não-lig}} = \sum_{i<j} \left[ 4\epsilon_{ij} \left( \left( \frac{\sigma_{ij}}{r_{ij}} \right)^{12} - \left( \frac{\sigma_{ij}}{r_{ij}} \right)^{6} \right) + \frac{q_i q_j}{4 \pi \epsilon_0 r_{ij}} \right]$$

Os parâmetros ($k_b$, $r_0$, $k_\theta$, $V_n$, $\epsilon$, $\sigma$, $q$) vêm de dados experimentais e de cálculos quânticos de referência. Cada família de campo de força é um conjunto próprio desses parâmetros, ajustado para um tipo de sistema: AMBER, CHARMM, OPLS e GROMOS para biomoléculas; MMFF94, UFF e GAFF para moléculas orgânicas pequenas.

Duas limitações que precisam ficar claras: um campo de força clássico **não quebra nem forma ligações** (a mola harmônica nunca solta) e **não descreve transferência de carga nem polarização** nas versões usuais. Para reação química, você precisa de mecânica quântica, que é o assunto da Parte 03.

O gráfico abaixo mostra a forma funcional de cada termo.

````python
# @title <font color='green'> 3.7.1 — A forma de cada termo do campo de força
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(2, 2, figsize=(11, 7.5))

# ---- estiramento de ligação: harmônico -------------------------------------
r0, kb = 1.53, 300.0  # C-C sp3: 1.53 A, ~300 kcal/mol/A^2
r = np.linspace(1.1, 2.0, 400)
E_stretch = 0.5 * kb * (r - r0) ** 2

ax[0, 0].plot(r, E_stretch, lw=2.4, color="#1a9850")
ax[0, 0].axvline(r0, ls="--", lw=1.1, color="gray")
ax[0, 0].set_ylim(-1.5, 33)
ax[0, 0].set_title(
    r"Estiramento:  $\frac{1}{2}k_b(r-r_0)^2$", fontsize=12, pad=12
)
ax[0, 0].set_xlabel("Distância de ligação r (Å)")
ax[0, 0].set_ylabel("E (kcal/mol)")
ax[0, 0].text(
    r0 + 0.03,
    22,
    r"$r_0$ = 1,53 Å",
    fontsize=9.5,
    color="gray",
    fontweight="bold",
)

# ---- deformação angular: harmônico -----------------------------------------
th0, kth = 109.5, 0.02  # tetraédrico
th = np.linspace(85, 135, 400)
E_angle = 0.5 * kth * (th - th0) ** 2

ax[0, 1].plot(th, E_angle, lw=2.4, color="#4575b4")
ax[0, 1].axvline(th0, ls="--", lw=1.1, color="gray")
ax[0, 1].set_ylim(-0.3, 6.8)
ax[0, 1].set_title(
    r"Angular:  $\frac{1}{2}k_\theta(\theta-\theta_0)^2$", fontsize=12, pad=12
)
ax[0, 1].set_xlabel("Ângulo θ (graus)")
ax[0, 1].set_ylabel("E (kcal/mol)")
ax[0, 1].text(
    th0 + 1.2,
    4.5,
    r"$\theta_0$ = 109,5°",
    fontsize=9.5,
    color="gray",
    fontweight="bold",
)

# ---- torção: série de Fourier ----------------------------------------------
phi = np.linspace(-180, 180, 600)
V3, V1 = 1.4, 1.4
E_tor = 0.5 * V3 * (1 + np.cos(np.radians(3 * phi)))

ax[1, 0].plot(phi, E_tor, lw=2.4, color="#f46d43")
ax[1, 0].set_ylim(-0.1, 2.0)
ax[1, 0].set_title(
    r"Torção:  $\sum_n \frac{1}{2}V_n[1+\cos(n\phi-\gamma)]$",
    fontsize=12,
    pad=12,
)
ax[1, 0].set_xlabel("Ângulo diedro φ (graus)")
ax[1, 0].set_ylabel("E (kcal/mol)")
ax[1, 0].set_xticks([-180, -120, -60, 0, 60, 120, 180])
ax[1, 0].text(
    -175,
    1.75,
    "Os dois mínimos do etano (alternada) e \ntrês máximos (eclipsada)",
    fontsize=8.5,
    color="#555555",
)

# ---- não ligadas: Lennard-Jones e Coulomb ----------------------------------
d = np.linspace(2.8, 8.0, 500)
eps, sig = 0.15, 3.4  # carbono, aproximadamente
E_lj = 4 * eps * ((sig / d) ** 12 - (sig / d) ** 6)
E_coul = 332.06 * (0.4 * -0.4) / (1.0 * d)  # cargas +0,4 e -0,4 no vácuo

ax[1, 1].plot(d, E_lj, lw=2.4, color="#762a83", label="Lennard-Jones (vdW)")
ax[1, 1].plot(
    d, E_coul, lw=2.4, color="#c51b7d", ls="--", label="Coulomb (q = ±0,4)"
)
ax[1, 1].axhline(0, color="k", lw=0.8)
ax[1, 1].set_ylim(-2.2, 1.4)
ax[1, 1].set_title(
    "Não ligadas: van der Waals + eletrostática", fontsize=12, pad=12
)
ax[1, 1].set_xlabel("Distância entre átomos (Å)")
ax[1, 1].set_ylabel("E (kcal/mol)")
ax[1, 1].legend(fontsize=9, loc="upper right")

for a in ax.ravel():
  a.grid(alpha=0.25)

plt.tight_layout()
plt.show()

print(
    "Note a assimetria dos dois primeiros: a mola harmônica sobe para sempre."
)
print(
    "Um campo de força clássico não quebra ligação. Para isso, a necessidade da mecânica"
    " quântica (Parte 03)."
)
````

## <font color='magenta'>**3.8 Onde a modelagem molecular é usada (opcional)**</font>

Uma leitura de três minutos sobre o alcance da área, para quem quiser situar o próprio trabalho.

**Descoberta de fármacos.** É a aplicação mais visível. Docking e dinâmica molecular estimam como um candidato se liga ao alvo; modelos QSAR e de ADMET filtram o que provavelmente vai falhar em absorção, metabolismo ou toxicidade antes de alguém sintetizar. O ganho não é substituir o experimento — é decidir quais dos cem mil compostos merecem os dez ensaios que você consegue pagar.

**Ciência de materiais.** Simulação de resistência mecânica e reatividade de superfície. Polímeros, nanomateriais, materiais para baterias e células solares. Boa parte do trabalho recente de triagem de materiais combina DFT com aprendizado de máquina.

## <font color='green'>**4.6 Energia com o campo de força MMFF94**</font>

Fechando o ciclo do módulo 4: uma SMILES entra, uma geometria 3D é gerada, e o campo de força MMFF94 devolve a energia daquela conformação.

Um aviso sobre o número que sai daqui. Energia de campo de força **não tem significado absoluto**. Ela só vale para comparar conformações da mesma molécula. Comparar a energia MMFF94 de duas moléculas diferentes não diz nada, porque os termos de referência são outros.

````python
#@title <font color='green'> 4.6.1 — Energia MMFF94 a partir de uma SMILES

from rdkit import Chem
from rdkit.Chem import AllChem

def calcula_energia_MMFF94(smiles):
    """
    Calcula a energia da molécula no campo de força MMFF94 usando RDKit.

    Parâmetro:
    smiles (str): Representação SMILES da molécula.

    Retorno:
    float: Energia da molécula (kcal/mol) ou None se MMFF não estiver disponível.
    """
    # Gerar a molécula a partir do SMILES
    mol = Chem.MolFromSmiles(smiles)

    # Adicionar hidrogênios explícitos
    mol = Chem.AddHs(mol)

    # Gerar uma conformação inicial
    AllChem.EmbedMolecule(mol, AllChem.ETKDG())

    # Verificar se os parâmetros MMFF estão disponíveis para a molécula
    mmff_props = AllChem.MMFFGetMoleculeProperties(mol)
    if mmff_props is None:
        print(f"Parâmetros MMFF não disponíveis para a molécula: {smiles}")
        return None

    # Criar o campo de força MMFF94
    ff = AllChem.MMFFGetMoleculeForceField(mol, mmff_props)

    # Calcular a energia da molécula em MMFF94
    energia = ff.CalcEnergy()

    return energia

# Exemplo: Calcular energia do benzeno (SMILES: "c1ccccc1")
smiles_molecula = "c1ccccc1"  # Benzeno
energia = calcula_energia_MMFF94(smiles_molecula)

if energia is not None:
    print(f"Energia da molécula ({smiles_molecula}) em MMFF94: {energia:.2f} kcal/mol")
````

### Detalhamento Etapa por Etapa

* **`Chem.MolFromSmiles(smiles)`**: Converte a string SMILES em um objeto molecular do RDKit (`Mol`) armazenado em memória.
* **`Chem.AddHs(mol)`**: Adiciona explicitamente os átomos de hidrogênio. Estruturas SMILES omitem hidrogênios apolares por padrão, mas cálculos de campos de força exigem todos os átomos para avaliar interações eletrostáticas, de van der Waals e termos torcionais.
* **`AllChem.EmbedMolecule(mol, AllChem.ETKDG())`**: Gera as coordenadas $3\text{D}$ iniciais da molécula utilizando o algoritmo **ETKDG** (*Experimental Torsion-angle preference with Knowledge-based Distance Geometry*).
* **`AllChem.MMFFGetMoleculeProperties(mol)`**: Atribui os tipos atômicos e parâmetros de carga do MMFF94 à molécula. Se a estrutura contiver elementos não suportados pelo campo de força (como certos metais ou átomos com hibridização atípica), a função retorna `None`.
* **`AllChem.MMFFGetMoleculeForceField(mol, mmff_props)`**: Instancia o objeto do campo de força vinculando as equações matemáticas do MMFF94 à geometria $3\text{D}$ gerada.
* **`ff.CalcEnergy()`**: Avalia e retorna a energia potencial total da conformação atual em $\text{kcal/mol}$.

## <font color='green'>**5.1 Teoria: descritores, filtros e métricas de eficiência**</font>

Antes de calcular, vale saber o que se está calculando. Cinco páginas curtas que sustentam tudo o que vem nas seções 5.2 a 5.8.

### <font color='green'>**5.1.1 Descritores moleculares**</font>

Descritor é qualquer número derivado da estrutura. A organização mais usada é por dimensionalidade da representação de origem.

| Dimensão | O que descreve | Exemplos |
|---|---|---|
| 0D | contagens independentes da conectividade | massa molecular, número de átomos |
| 1D | listas de fragmentos e grupos funcionais | contagem de anéis, doadores e aceptores de ligação de hidrogênio |
| 2D | propriedades do grafo molecular | índices topológicos |
| 3D | geometria e campos no espaço | momentos de inércia, campos estéricos e eletrostáticos |
| 4D | conjuntos conformacionais | descritores sobre conformações |

Os **índices topológicos** merecem menção porque foram a primeira ponte entre teoria de grafos e química. O índice de Wiener (1947) é a soma das distâncias topológicas entre todos os pares de átomos,

$$W = \frac{1}{2}\sum_i\sum_j d_{ij}$$

e correlaciona surpreendentemente bem com pontos de ebulição de alcanos. O índice de conectividade de Randić (1975) pondera cada ligação pelos graus dos átomos envolvidos,

$$\chi = \sum_{\text{ligações}} \left(\delta_i\,\delta_j\right)^{-1/2}$$

capturando o grau de ramificação da molécula.

Os descritores **físico-químicos** — $\log P$, $\text{p}K_a$, área polar superficial, refratividade molar, polarizabilidade — são os mais interpretáveis e os mais usados em química medicinal.

Os descritores **quânticos** conectam esta parte à Parte 03 do minicurso: energias de orbitais moleculares de fronteira, dureza, índice de eletrofilicidade, momento de dipolo, cargas atômicas, superfícies de potencial eletrostático. São particularmente informativos quando o mecanismo envolve reatividade, como em inibidores covalentes.

### <font color='magenta'>**5.1.2 Impressões digitais**</font>

Descritores reduzem a molécula a poucos números interpretáveis. As **impressões digitais** fazem o oposto: registram presença de subestruturas num vetor de bits longo. Dividem-se em dois tipos.

As **baseadas em chaves**, como as MACCS, têm cada posição correspondendo a uma pergunta estrutural predefinida, do tipo "existe um anel aromático de seis membros com nitrogênio?". São interpretáveis, e limitadas ao dicionário fixo.

As **baseadas em resumo criptográfico**, como as circulares ECFP/Morgan (Rogers e Hahn, 2010), enumeram subestruturas e as mapeiam por função de resumo em posições de um vetor de bits de tamanho fixo. Cobrem qualquer tipo de estrutura, ao custo de **colisões**: duas subestruturas diferentes podem cair na mesma posição do bit, e a posição deixa de ter interpretação única.

A escolha envolve um compromisso explícito. Vetores curtos colidem mais (512 bits) e perdem resolução; vetores longos (4096 bits) são esparsos e caros. E o raio nas circulares controla o tamanho do ambiente químico capturado: raio 2 vê a vizinhança até duas ligações, o que corresponde aproximadamente a um grupo funcional com seu entorno imediato.

Uma implicação para leitura de resultados: atribuir significado químico a um bit específico exige verificar quais subestruturas efetivamente o ativam. Por causa das colisões, um bit isolado não tem interpretação garantida.

### <font color='green'>**5.1.3 Regras de propriedade: o que elas dizem e o que não dizem**</font>

A **regra dos cinco** (Lipinski e colaboradores, 1997) e os critérios de Veber e colaboradores (2002) delimitam faixas de propriedades associadas à absorção oral. Para fragmentos, a **regra dos três** (Congreve e colaboradores, 2003) usa limiares proporcionalmente menores.

Duas ressalvas precisam acompanhar sempre essas regras.

Elas descrevem **absorção oral**, não atividade. Uma molécula pode violar tudo e ser excelente inibidor; apenas não será comprimido.

E foram derivadas de compostos que chegaram a ensaio clínico nos anos 1990. Produtos naturais, antibióticos, peptídeos e degradadores bifuncionais as violam rotineiramente. Vamos ver isso acontecer na seção 6.7, com compostos que qualquer químico reconhece como fármacos e que a regra reprova.

### <font color='magenta'>**5.1.4 Métricas de eficiência**</font>

Comparar potências brutas favorece moléculas grandes e gordurosas, que ganham afinidade por contatos inespecíficos. As métricas de eficiência corrigem isso.

A **eficiência do ligante** (Hopkins, Groom e Alex, 2004) normaliza a energia livre de ligação pelo número de átomos pesados:

$$\text{LE} = \frac{-\Delta G^\circ}{N_{\text{pesados}}} \;\approx\; \frac{1{,}37\; pIC_{50}}{N_{\text{pesados}}}$$

onde o fator numérico vem de $RT\ln 10$ a 298 K, em kcal/mol.

A **eficiência lipofílica** (Leeson e Springthorpe, 2007) desconta a contribuição inespecífica da lipofilicidade:

$$\text{LLE} = pIC_{50} - \log P$$

O valor prático dessas métricas é orientar a otimização. Um substituinte que aumenta a potência em dez vezes ao custo de duas unidades de $\log P$ piorou a molécula, mesmo tendo melhorado o número que aparece no ensaio.

### <font color='magenta'>**5.1.5 Alertas estruturais e desejabilidade**</font>

Os **PAINS (*Pan-Assay Interference Compounds*)** (Baell e Holloway, 2010) catalogam subestruturas que aparecem como ativas em muitos ensaios não relacionados, por mecanismos como interferência óptica, reatividade inespecífica, quelação e formação de agregados coloidais. Esses compostos não se ligam especificamente ao alvo terapêutico de interesse. Em vez disso, eles "enganam" o ensaio experimental por meio de mecanismos artefatosos.

São escritos em SMARTS e aplicados como filtro.

Cabe aqui a advertência que a literatura recente enfatiza: os padrões PAINS foram derivados de um conjunto específico de ensaios de proximidade luminescente, e usá-los como veto automático descarta química legítima. O uso correto é como sinal de alerta que motiva controle experimental, não como critério eliminatório.

O **QED (*Quantitative Estimate of Drug-likeness*)** (Bickerton e colaboradores, 2012) agrega várias propriedades num único índice, aplicando funções de desejabilidade $d_i \in [0,1]$ a cada propriedade e tomando a média geométrica:

$$\text{QED} = \exp\left(\frac{1}{n}\sum_{i=1}^{n}\ln d_i\right)$$

A média geométrica é escolha deliberada. Como qualquer $d_i$ próximo de zero derruba o produto inteiro, o índice não permite que uma propriedade excelente compense outra desastrosa, comportamento que a média aritmética permitiria.

## <font color='green'>**5.2 Lendo SMILES com o RDKit**</font>

````python
#@title <font color='green'> 5.2.1 — Lendo SMILES
# Vamos ler três fármacos e ver o que o RDKit faz com eles.

exemplos = {
    "Aspirina":    "CC(=O)Oc1ccccc1C(=O)O",
    "Paracetamol": "CC(=O)Nc1ccc(O)cc1",
    "Ibuprofeno":  "CC(C)Cc1ccc(cc1)C(C)C(=O)O",
}

for nome, smi in exemplos.items():
    mol = Chem.MolFromSmiles(smi)          # texto -> objeto molécula
    print(f"{nome}")
    print(f"   SMILES de entrada : {smi}")
    print(f"   SMILES canônica   : {Chem.MolToSmiles(mol)}")
    print(f"   Fórmula           : {rdmd.CalcMolFormula(mol)}")
    print(f"   Átomos pesados    : {mol.GetNumHeavyAtoms()}  (sem contar H implícitos)")
    print()
````

## <font color='green'>**5.5 Descritores: números que resumem a molécula**</font>

Um descritor é qualquer número calculado a partir da estrutura. O RDKit traz mais de duzentos. A maioria você nunca vai usar. Os que aparecem em praticamente todo trabalho de descoberta de fármacos são meia dúzia:

- **MW** — massa molar. Moléculas grandes difundem mal e costumam ter problemas de solubilidade.
- **cLogP** — logaritmo do coeficiente de partição octanol/água, calculado. Mede lipofilicidade. Muito baixo, a molécula não atravessa membrana; muito alto, ela fica presa na gordura e se liga a tudo.
- **TPSA** — área de superfície polar topológica. Soma das contribuições de átomos de N e O (e seus H). É o melhor preditor simples de permeabilidade que existe.
- **HBD / HBA** — doadores e aceptores de ligação de hidrogênio. Cada ligação de H que a molécula faz com a água precisa ser desfeita para atravessar a membrana. Custa energia.
- **RotB** — ligações rotacionáveis. Mais flexibilidade significa mais entropia perdida ao ligar num alvo.

A **regra dos cinco**, publicada por Christopher Lipinski em 1997, diz que compostos com MW > 500, cLogP > 5, HBD > 5 ou HBA > 10 tendem a ter absorção oral ruim. Ela não é lei — é uma observação estatística sobre fármacos que passaram na fase II. Antibióticos, produtos naturais e transportados ativamente violam a regra e funcionam mesmo assim.

````python
#@title <font color='green'> 5.5.1 — Calculando descritores
# Calculando o painel de descritores para os três fármacos.

def descritores(mol):
    '''Painel mínimo de descritores físico-químicos.'''
    return {
        "MW":    Descriptors.MolWt(mol),
        "cLogP": Crippen.MolLogP(mol),
        "TPSA":  rdmd.CalcTPSA(mol),
        "HBD":   rdmd.CalcNumHBD(mol),
        "HBA":   rdmd.CalcNumHBA(mol),
        "RotB":  rdmd.CalcNumRotatableBonds(mol),
        "Anéis": rdmd.CalcNumRings(mol),
        "FrCsp3": rdmd.CalcFractionCSP3(mol),
    }

df_desc = pd.DataFrame(
    [descritores(m) for m in mols],
    index=list(exemplos.keys())
).round(2)

df_desc
````

## <font color='green'>**5.6 Regras de Lipinski e Veber**</font>

````python
#@title <font color='green'> 5.6.1 — Regras de Lipinski e Veber
# Aplicando a regra dos cinco de Lipinski e de Veber.

def lipinski(mol):
    v = []
    if Descriptors.MolWt(mol) > 500:      v.append("MW>500")
    if Crippen.MolLogP(mol) > 5:          v.append("cLogP>5")
    if rdmd.CalcNumHBD(mol) > 5:          v.append("HBD>5")
    if rdmd.CalcNumHBA(mol) > 10:         v.append("HBA>10")
    return v

def veber(mol):
    '''Critérios de Veber (2002), voltados a biodisponibilidade oral.'''
    v = []
    if rdmd.CalcNumRotatableBonds(mol) > 10: v.append("RotB>10")
    if rdmd.CalcTPSA(mol) > 140:             v.append("TPSA>140")
    return v

for nome, mol in zip(exemplos.keys(), mols):
    li, ve = lipinski(mol), veber(mol)
    status = "passa" if not li and not ve else "violações: " + ", ".join(li + ve)
    print(f"{nome:14s} {status}")
````

Os três passam, como era de esperar — são fármacos orais consagrados. O teste fica mais interessante quando você joga na regra um composto que não é fármaco.

## <font color='green'>**5.7 Quando a regra falha**</font>

````python
#@title <font color='green'> 5.7.1 — Quando a regra falha
# Contraste: quatro moléculas que a regra dos cinco reprova,
# e três delas são medicamentos usados clinicamente.

fora_da_regra = {
    "Ciclosporina A": "CC[C@H]1C(=O)N(CC(=O)N([C@H](C(=O)N[C@H](C(=O)N([C@H](C(=O)N[C@H](C(=O)N[C@@H](C(=O)N([C@H](C(=O)N([C@H](C(=O)N([C@H](C(=O)N([C@H](C(=O)N1)[C@@H]([C@H](C)C/C=C/C)O)C)C(C)C)C)CC(C)C)C)CC(C)C)C)C)C)CC(C)C)C)C(C)C)CC(C)C)C)C",
    "Vancomicina":    "C[C@H]1[C@H]([C@@](C[C@@H](O1)O[C@@H]2[C@H]([C@@H]([C@H](O[C@H]2OC3=C4C=C5C=C3OC6=C(C=C(C=C6)[C@H]([C@H](C(=O)N[C@H](C(=O)N[C@H]5C(=O)N[C@@H]7C8=CC(=C(C=C8)O)C9=C(C=C(C=C9O)O)[C@H](NC(=O)[C@H]([C@@H](C1=CC(=C(O4)C=C1)Cl)O)NC7=O)C(=O)O)CC(=O)N)NC(=O)[C@@H](CC(C)C)NC)O)Cl)CO)O)O)(C)N)O",
    "Atorvastatina":  "CC(C)C1=C(C(=C(N1CC[C@H](C[C@H](CC(=O)O)O)O)C2=CC=C(C=C2)F)C3=CC=CC=C3)C(=O)NC4=CC=CC=C4",
    "Sacarose":       "OCC1OC(OC2(CO)OC(CO)C(O)C2O)C(O)C(O)C1O",
}

linhas = []
for nome, smi in fora_da_regra.items():
    m = Chem.MolFromSmiles(smi)
    d = descritores(m)
    d["Violações"] = ", ".join(lipinski(m) + veber(m)) or "nenhuma"
    linhas.append(pd.Series(d, name=nome))

pd.DataFrame(linhas).round(1)
````

Ciclosporina e vancomicina violam quase tudo e são medicamentos essenciais. A ciclosporina é absorvida por via oral apesar de ter massa 1202 Da — porque forma ligações de hidrogênio internas que escondem os grupos polares quando ela está na membrana. A regra de Lipinski não enxerga isso, porque é topológica e não sabe nada sobre conformação.

A lição importante a ser retirada deste exercício: descritores são um filtro de triagem, não um veredito. Eles ordenam prioridades quando você tem cem mil compostos e tempo/recurso para testar dez na bancada. Usados como critério de exclusão absoluto, eles descartam justamente as moléculas interessantes.

# <font color='green'> **6. Exercício 2 — Espaço químico**

Agora saímos de três moléculas para sessenta. A pergunta muda: dado um conjunto grande de compostos, como você mede o quanto dois deles se parecem, como agrupa os semelhantes, e como decide quais merecem atenção?

Isso é o coração da triagem virtual baseada em ligante — a estratégia que você usa quando conhece moléculas ativas mas não tem a estrutura do alvo. Situação bem comum com produtos naturais, aliás, que é o que muita gente da RENORBIO trabalha.

## <font color='magenta'>**6.1 Teoria: similaridade, busca e curadoria**</font>

O módulo anterior transformou a molécula em números. Este a transforma em vizinhança: quem se parece com quem, sob qual medida, e o que isso autoriza concluir.

### <font color='magenta'>**6.1.1 O formalismo da similaridade**</font>

Definida a representação, a proximidade entre moléculas vira uma questão matemática bem posta. Para vetores binários, sejam $a$ e $b$ os números de bits ligados em cada molécula e $c$ o número de bits ligados em ambas:

$$T = \frac{c}{a+b-c}, \qquad D = \frac{2c}{a+b}, \qquad C = \frac{c}{\sqrt{ab}}$$

O coeficiente de **Tanimoto** $T$ é o padrão da área de quimioinformática; $D$ é o de Dice e $C$ o cosseno.

Uma relação que evita discussões inúteis: Dice e Tanimoto estão ligados por

$$D = \frac{2T}{1+T}$$

que é monotonicamente crescente. Como a transformação preserva ordem, os dois coeficientes produzem exatamente o mesmo ranking. Na vasta maioria dos casos, escolher entre eles muda os valores absolutos, não a ordenação, e portanto não muda o resultado de uma busca por similaridade.

**Tanimoto é um índice de similaridade.** Para vetores binários, a distância $d = 1 - T$ satisfaz a desigualdade triangular, e portanto define uma métrica legítima no espaço químico (Lipkus, 1999). Não é detalhe formal: é o que autoriza usar algoritmos de agrupamento, indexação métrica e busca por vizinhos mais próximos que pressupõem espaço métrico. Para representações contínuas a propriedade não vale em geral, e a transposição descuidada de algoritmos pode falhar em silêncio.

## <font color='green'>**6.4 Impressões digitais moleculares**</font>

Descritores reduzem a molécula a poucos números. *Fingerprints* fazem outra coisa: transformam a estrutura num vetor binário de milhares de posições, onde cada bit indica a presença de um fragmento.

O tipo mais usado é o **Morgan** (também chamado ECFP, *extended connectivity fingerprint*). O algoritmo percorre cada átomo, olha a vizinhança até um certo raio, gera um identificador para aquele ambiente químico e liga o bit correspondente. Raio 2 significa que ele enxerga até dois átomos de distância.

Com dois vetores binários, você compara com o **coeficiente de Tanimoto**: número de bits em comum dividido pelo número de bits ligados em pelo menos um dos dois. Vai de 0 a 1.

A regra prática é que Tanimoto acima de 0,85 indica atividade biológica parecida. Essa regra tem exceções suficientes para ser vista com cautela — vem de um estudo de 1996 com um conjunto específico e não generaliza bem. Deve-se tratar como indício, não como prova.

````python
#@title <font color='green'> 6.4.1 — Gerando fingerprints
# Gerando os fingerprints.

gerador = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)
fps = [gerador.GetFingerprint(m) for m in quimioteca.mol]

print(f"{len(fps)} fingerprints de {fps[0].GetNumBits()} bits")
print(f"Bits ligados na primeira molécula ({quimioteca.nome[0]}): "
      f"{fps[0].GetNumOnBits()}")

# O vetor é esparso: pouquíssimos bits ligados entre 2048.
ocupacao = [fp.GetNumOnBits() / fp.GetNumBits() for fp in fps]
print(f"Ocupação média do vetor: {100*np.mean(ocupacao):.1f}%")
````

## <font color='green'>**6.5 Busca por similaridade**</font>

````python
#@title <font color='green'> 6.5.1 — Busca por similaridade
# Busca por similaridade: dado um composto de referência, quem se parece com ele?
# É assim que funciona uma triagem virtual baseada em ligante.

def buscar_similares(nome_ref, n=6):
    i = quimioteca.index[quimioteca.nome == nome_ref][0]
    sims = DataStructs.BulkTanimotoSimilarity(fps[i], fps)
    ordem = np.argsort(sims)[::-1][1:n+1]       # [1:] pula a própria molécula
    return pd.DataFrame({
        "composto": quimioteca.nome[ordem].values,
        "classe":   quimioteca.classe[ordem].values,
        "Tanimoto": np.round([sims[j] for j in ordem], 3),
    })

for ref in ["Fluoxetina", "Diazepam", "PenicilinaG"]:
    print(f"\nMais parecidos com {ref}:")
    print(buscar_similares(ref, 5).to_string(index=False))
````

Os valores de Tanimoto vieram baixos, quase todos abaixo de 0,25. Isso não é erro — é informação. Fármacos aprovados ocupam um espaço químico muito disperso, porque foram descobertos por caminhos independentes ao longo de um século. Numa quimioteca combinatória de verdade, com dezenas de milhares de análogos, você veria valores bem mais altos e clusters densos.

Vale prestar atenção num caso: diazepam, alprazolam, midazolam e lorazepam são todos benzodiazepínicos e aparecem próximos entre si. O fingerprint captou a classe estrutural sem que ninguém tenha dito a ele o que é um benzodiazepínico.

## <font color='green'>**6.6 Matriz de similaridade**</font>

````python
#@title <font color='green'> 6.6.1 — Matriz de similaridade
# Matriz de similaridade completa e mapa de calor.

n = len(fps)
matriz = np.ones((n, n))
for i in range(1, n):
    s = DataStructs.BulkTanimotoSimilarity(fps[i], fps[:i])
    matriz[i, :i] = s   # a matriz deve ser simétrica, por isso essa imposição !
    matriz[:i, i] = s

# Ordeno por classe terapêutica para que blocos apareçam, se existirem
ordem = np.argsort(quimioteca.classe.values, kind="stable")

fig, ax = plt.subplots(figsize=(8, 6.8))
im = ax.imshow(matriz[np.ix_(ordem, ordem)], cmap="viridis", vmin=0, vmax=0.6)
ax.set_xticks(range(n)); ax.set_yticks(range(n))
ax.set_xticklabels(quimioteca.nome.values[ordem], rotation=90, fontsize=5)
ax.set_yticklabels(quimioteca.nome.values[ordem], fontsize=5)
ax.grid(False)
plt.colorbar(im, label="Tanimoto")
ax.set_title("Similaridade entre pares (ordenado por classe terapêutica)")
plt.tight_layout(); plt.show()
````

## <font color='magenta'>**6.7 Agrupamento de Butina**</font>

````python
#@title <font color='magenta'> 6.7.1 — Agrupamento de Butina
# Agrupamento de Butina: um método clássico em quimioinformática.
# Ele escolhe o composto com mais vizinhos dentro do corte, forma um cluster,
# remove todos eles e repete. Simples e determinístico.

CORTE = 0.65   # distância = 1 - Tanimoto. Vizinhos com Tc > 0.35 entram juntos.

distancias = []
for i in range(1, n):
    distancias.extend([1 - x for x in DataStructs.BulkTanimotoSimilarity(fps[i], fps[:i])])

clusters = Butina.ClusterData(distancias, n, CORTE, isDistData=True)

print(f"{len(clusters)} clusters para {n} compostos (corte de distância {CORTE})")
print(f"Compostos isolados (cluster de 1): {sum(1 for c in clusters if len(c) == 1)}")
print("\nClusters com mais de um membro:")
for c in clusters:
    if len(c) > 1:
        print("  ", ", ".join(quimioteca.nome[list(c)].values))
````

Quarenta e nove clusters para sessenta compostos. A quimioteca é quase toda *singleton* — cada fármaco é um caso à parte. Os grupos que sobraram fazem sentido químico: os benzodiazepínicos se juntam, os beta-lactâmicos se juntam.

Se você fizer isso com uma biblioteca comercial de triagem, o padrão inverte: poucos clusters enormes, porque o conjunto de fármacos foi construído aqui como um *toy-dataset*.

## <font color='green'>**6.8 Mapa do espaço químico**</font>

````python
#@title <font color='green'> 6.8.1 — Mapa do espaço químico
# Mapa do espaço químico. Faço duas projeções para comparar.

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# (a) PCA sobre os fingerprints binários
X_fp = np.zeros((n, 2048), dtype=np.int8)
for i, fp in enumerate(fps):
    arr = np.zeros((2048,), dtype=np.int8)
    DataStructs.ConvertToNumpyArray(fp, arr)
    X_fp[i] = arr
pca_fp = PCA(n_components=2).fit(X_fp)
proj_fp = pca_fp.transform(X_fp)

# (b) PCA sobre descritores físico-químicos padronizados
cols = ["MW", "cLogP", "TPSA", "HBD", "HBA", "RotB", "FrCsp3"]
X_desc = StandardScaler().fit_transform(quimioteca[cols].values)
pca_desc = PCA(n_components=2).fit(X_desc)
proj_desc = pca_desc.transform(X_desc)

fig, axes = plt.subplots(1, 2, figsize=(12.5, 5))
for ax, proj, pca, titulo in [
        (axes[0], proj_fp,   pca_fp,   "PCA sobre fingerprints ECFP4"),
        (axes[1], proj_desc, pca_desc, "PCA sobre descritores físico-químicos")]:
    for rot, cor, lab in [(1, "#2c7fb8", "atravessa BHE"), (0, "#e6842a", "não atravessa")]:
        sel = quimioteca.bbb == rot
        ax.scatter(proj[sel.values, 0], proj[sel.values, 1], c=cor, s=55,
                   alpha=0.8, edgecolor="k", linewidth=0.4, label=lab)
    var = pca.explained_variance_ratio_
    ax.set_xlabel(f"PC1 ({100*var[0]:.1f}%)")
    ax.set_ylabel(f"PC2 ({100*var[1]:.1f}%)")
    ax.set_title(titulo)
    ax.legend(fontsize=9)
plt.tight_layout(); plt.show()
````

Ao comparar as variâncias acumuladas do PCA, observa-se que os dois primeiros componentes explicam apenas ~10% da variância dos *fingerprints* binários, contra ~75% dos descritores físico-químicos. Esse contraste reflete a diferença fundamental de **dimensionalidade intrínseca** entre as duas representações:

* **Descritores Físico-Químicos:** Variáveis contínuas e correlacionadas (como MW, TPSA e LogP) mapeiam diretamente as propriedades biofísicas que governam a difusão passiva pela Barreira Hematoencefálica (BHE). A colinearidade entre essas métricas permite que o PCA comprima a informação em poucas dimensões, gerando agrupamentos nítidos por classe de permeabilidade.
* **Fingerprints Estruturais (ECFP4):** Representam 2048 dimensões binárias e altamente esparsas. Como os subfragmentos moleculares estão distribuídos em eixos ortogonais independentes, qualquer projeção linear (como o PCA) perde a maior parte da variabilidade global ao tentar comprimi-la em duas dimensões.

Essa análise desmistifica uma interpretação equivocada comum no mapeamento do espaço químico:

1. **Limitação do PCA em Fingerprints:** Devido à baixa variância explicada (~10%), as distâncias euclidianas no gráfico 2D do PCA não refletem as distâncias de Tanimoto do espaço original.
2. **Diferença para Métodos Não-Lineares (t-SNE / UMAP):** Ao contrário do PCA, algoritmos como t-SNE e UMAP não tentam maximizar a variância linear total. Eles otimizam a preservação da **topologia de vizinhança local ($k$-NN)**. Embora sejam excelentes para agrupar moléculas análogas, eles distorcem inevitavelmente as **distâncias globais** entre *clusters* distantes.

**Conclusão:** Mapeamentos 2D do espaço químico — sejam por PCA, t-SNE ou UMAP — são ferramentas qualitativas valiosas para identificação de padrões e navegação exploratória. No entanto, não devem ser utilizados isoladamente para tirar conclusões quantitativas sobre a similaridade química global.

````python
#@title <font color='magenta'> 6.9.1 — Filtros de alerta estrutural
# Alertas estruturais: PAINS e Brenk.
# PAINS = pan-assay interference compounds, fragmentos que dão falso positivo
# em muitos ensaios (por fluorescência, agregação, reatividade inespecífica).

def catalogo(tipo):
    p = FilterCatalogParams()
    p.AddCatalog(getattr(FilterCatalogParams.FilterCatalogs, tipo))
    return FilterCatalog(p)

cat_pains, cat_brenk = catalogo("PAINS"), catalogo("BRENK")

alertas = []
for _, r in quimioteca.iterrows():
    p = cat_pains.GetFirstMatch(r.mol)
    b = cat_brenk.GetFirstMatch(r.mol)
    if p or b:
        alertas.append({"composto": r.nome, "classe": r.classe,
                        "PAINS": p.GetDescription() if p else "—",
                        "Brenk": b.GetDescription() if b else "—"})

df_alertas = pd.DataFrame(alertas)
print(f"{len(df_alertas)} dos {n} fármacos aprovados disparam algum alerta\n")
df_alertas
````

Este resultado é bem importante !

Morfina, carbamazepina, dopamina e sulfassalazina disparam alertas. São medicamentos aprovados, alguns na lista de medicamentos essenciais da OMS, e um filtro automático os descartaria da sua triagem.

O que está acontecendo: PAINS foi derivado de compostos que interferiram em ensaios de AlphaScreen num conjunto de 93 mil moléculas. É um alerta sobre o **ensaio**, não sobre a molécula. Catecol dispara PAINS porque catecóis oxidam e reagem — o que é verdade da dopamina e é exatamente por isso que ela funciona como neurotransmissor.

Filtros servem para você olhar com atenção redobrada para o que foi marcado, não para deletar automaticamente. Um artigo de 2017 no *Journal of Medicinal Chemistry* estimou que aplicar PAINS cegamente descartaria uma fração relevante de compostos legítimos.

# <font color='green'> **7. Exercício 3 — QSAR com aprendizado de máquina**

Vamos construir um modelo que prevê se um composto atravessa a barreira hematoencefálica, usando só a estrutura.

O problema é real e relevante. A BHE é o motivo pelo qual boa parte dos candidatos a fármaco para doenças neurodegenerativas falha: a molécula funciona no ensaio bioquímico, funciona na célula, e não chega ao cérebro. Prever isso cedo economiza anos de esforços e custos.

QSAR — relação quantitativa estrutura-atividade — é a ideia de que propriedades biológicas são função da estrutura dos candidatos à fármacos, e que essa função pode ser aprendida a partir de exemplos. A formulação é dos anos 1960, com Hansch e Fujita. O que mudou desde então foram os algoritmos, não o conceito.

## <font color='green'>**7.1 Preparando o terreno**</font>

````python
#@title <font color='green'> 7.1.1 — Preparando o terreno
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import (train_test_split, StratifiedKFold,
                                     cross_val_score, cross_val_predict)
from sklearn.metrics import (roc_auc_score, roc_curve, balanced_accuracy_score,
                             confusion_matrix, ConfusionMatrixDisplay,
                             classification_report)
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.dummy import DummyClassifier

print("Módulos de aprendizado de máquina prontos.")
print(f"Conjunto: {len(quimioteca)} compostos, "
      f"{(quimioteca.bbb==1).sum()} positivos, {(quimioteca.bbb==0).sum()} negativos")
````

Note o desbalanceamento: 37 positivos contra 23 negativos. Não é ruim, mas é o suficiente para atrapalhar a criação do modelo. Um modelo inocente que responde "atravessa" para tudo já acerta 62% dos casos. Por isso não vamos usar acurácia como métrica — vamos usar acurácia balanceada e área sob a curva ROC, que não se deixam enganar por isso.

Sugestão de fonte bibliográfica:

- [Cheminformatics](https://chem.libretexts.org/Courses/Intercollegiate_Courses/Cheminformatics) — Curso online de quimioinformática acessível no portal [ChemLibreTexts](https://chem.libretexts.org/).

## <font color='green'>**7.2 Featurização**</font>

````python
#@title <font color='green'> 7.2.1 — Featurização
# Featurização: três representações diferentes da mesma molécula.

DESCRITORES = {
    "MW": Descriptors.MolWt, "cLogP": Crippen.MolLogP, "TPSA": rdmd.CalcTPSA,
    "HBD": rdmd.CalcNumHBD, "HBA": rdmd.CalcNumHBA,
    "RotB": rdmd.CalcNumRotatableBonds, "Anéis": rdmd.CalcNumRings,
    "AnéisArom": rdmd.CalcNumAromaticRings, "FrCsp3": rdmd.CalcFractionCSP3,
    "AtPesados": lambda m: m.GetNumHeavyAtoms(),
    "RefracMolar": Crippen.MolMR, "Heteroátomos": rdmd.CalcNumHeteroatoms,
}

X_descritores = pd.DataFrame(
    {nome: [f(m) for m in quimioteca.mol] for nome, f in DESCRITORES.items()})

gen_1024 = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=1024)

def para_vetor(mol):
    arr = np.zeros((1024,), dtype=np.int8)
    DataStructs.ConvertToNumpyArray(gen_1024.GetFingerprint(mol), arr)
    return arr

X_fingerprint = np.vstack([para_vetor(m) for m in quimioteca.mol])
X_combinado = np.hstack([X_descritores.values, X_fingerprint])

y = quimioteca.bbb.values

print(f"Descritores : {X_descritores.shape}")
print(f"Fingerprints: {X_fingerprint.shape}")
print(f"Combinado   : {X_combinado.shape}")
````

## <font color='green'>**7.3 Comparando representações**</font>

````python
#@title <font color='green'> 7.3.1 — Comparando representações
# Comparação das três representações, com validação cruzada.
# Uso 5 partições estratificadas: cada partição mantém a proporção de classes.

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

def avaliar(X, nome, modelo=None):
    modelo = modelo or RandomForestClassifier(
        n_estimators=400, random_state=42, class_weight="balanced", n_jobs=-1)
    auc = cross_val_score(modelo, X, y, cv=cv, scoring="roc_auc")
    bac = cross_val_score(modelo, X, y, cv=cv, scoring="balanced_accuracy")
    print(f"{nome:24s} AUC = {auc.mean():.3f} ± {auc.std():.3f}    "
          f"acurácia balanceada = {bac.mean():.3f} ± {bac.std():.3f}")
    return auc.mean()

print("Random Forest, validação cruzada 5x:\n")
avaliar(np.zeros((len(y), 1)), "linha de base (chute)", DummyClassifier(strategy="stratified"))
avaliar(X_descritores.values, "descritores (12)")
avaliar(X_fingerprint,        "fingerprint ECFP4 (1024)")
avaliar(X_combinado,          "descritores + ECFP4")
````

Doze descritores físico-químicos batem mil e vinte e quatro bits de fingerprint. Isso surpreende quem vem do aprendizado de máquina, onde a intuição é que mais variáveis ajudam.

A explicação tem duas partes. Primeiro, sessenta amostras não sustentam mil variáveis — o modelo decora em vez de aprender. Segundo, e mais interessante: a permeabilidade da BHE é governada por propriedades físico-químicas globais (lipofilicidade, polaridade, capacidade de fazer ligação de hidrogênio), não por fragmentos específicos. O descritor certo já codifica o mecanismo.

Isso é uma regra geral que vale a pena levar: fingerprints brilham quando a atividade depende de reconhecimento molecular específico, como afinidade por um alvo. Descritores brilham quando a propriedade é de transporte ou distribuição. Escolher a representação é modelagem, não pré-processamento.

## <font color='green'>**7.4 Treinando e testando**</font>

````python
#@title <font color='green'> 7.4.1 — Treinando e testando
# Treinando o modelo final e olhando o desempenho no conjunto de teste.

X_tr, X_te, y_tr, y_te, idx_tr, idx_te = train_test_split(
    X_descritores.values, y, np.arange(len(y)),
    test_size=0.25, stratify=y, random_state=42)

modelo = RandomForestClassifier(n_estimators=400, random_state=42,
                                class_weight="balanced", n_jobs=-1).fit(X_tr, y_tr)

prob = modelo.predict_proba(X_te)[:, 1]
pred = (prob >= 0.5).astype(int)

print(f"Treino: {len(y_tr)} compostos    Teste: {len(y_te)} compostos\n")
print(f"AUC ROC              : {roc_auc_score(y_te, prob):.3f}")
print(f"Acurácia balanceada  : {balanced_accuracy_score(y_te, pred):.3f}\n")
print(classification_report(y_te, pred,
                            target_names=["não atravessa", "atravessa"]))
````

## <font color='green'>**7.5 Diagnóstico do modelo**</font>

````python
#@title <font color='green'> 7.5.1 — Diagnóstico do modelo
# Curva ROC, matriz de confusão e importância das variáveis, lado a lado.

fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))

# ROC
fpr, tpr, _ = roc_curve(y_te, prob)
axes[0].plot(fpr, tpr, lw=2, color="#2c7fb8",
             label=f"AUC = {roc_auc_score(y_te, prob):.3f}")
axes[0].plot([0, 1], [0, 1], "k--", lw=1, label="chute aleatório")
axes[0].set_xlabel("taxa de falsos positivos")
axes[0].set_ylabel("taxa de verdadeiros positivos")
axes[0].set_title("Curva ROC")
axes[0].legend(fontsize=9)

# Matriz de confusão
ConfusionMatrixDisplay(confusion_matrix(y_te, pred),
                       display_labels=["não\natravessa", "atravessa"]).plot(
    ax=axes[1], cmap="Blues", colorbar=False)
axes[1].set_title("Matriz de confusão")
axes[1].grid(False)

# Importância das variáveis
imp = pd.Series(modelo.feature_importances_,
                index=X_descritores.columns).sort_values()
axes[2].barh(imp.index, imp.values, color="#7fcdbb", edgecolor="k", linewidth=0.4)
axes[2].set_xlabel("importância (Gini)")
axes[2].set_title("O que o modelo usa para decidir")

plt.tight_layout(); plt.show()
````

As variáveis mais importantes são TPSA, doadores de ligação de hidrogênio e cLogP. O modelo redescobriu, a partir de sessenta exemplos, o que a literatura de farmacocinética estabeleceu ao longo de décadas: para atravessar a BHE, uma molécula precisa ser pequena, moderadamente lipofílica e com pouca superfície polar.

## <font color='green'>**7.6 Prevendo compostos novos**</font>

````python
#@title <font color='green'> 7.6.1 — Prevendo compostos novos
# Aplicando o modelo a compostos que ele nunca viu.
# Treino agora com todos os 60 e prevejo moléculas novas.

modelo_final = RandomForestClassifier(n_estimators=400, random_state=42,
                                      class_weight="balanced", n_jobs=-1)
modelo_final.fit(X_descritores.values, y)

novos = {
    "Cafeína (controle +)":  "CN1C=NC2=C1C(=O)N(C)C(=O)N2C",
    "Galantamina":           "CN1CCC23C=CC(O)CC2Oc2c(OC)ccc(C1)c23",
    "Curcumina":             "COc1cc(/C=C/C(=O)CC(=O)/C=C/c2ccc(O)c(OC)c2)ccc1O",
    "Resveratrol":           "Oc1ccc(/C=C/c2cc(O)cc(O)c2)cc1",
    "Quercetina":            "O=c1c(O)c(-c2ccc(O)c(O)c2)oc2cc(O)cc(O)c12",
    "Artemisinina":          "CC1CCC2C(C)C(=O)OC3OC4(C)CCC1C32OO4",
    "Levofloxacino":         "CC1COc2c3n1cc(C(=O)O)c(=O)c3cc(F)c2N1CCN(C)CC1",
}

linhas = []
for nome, smi in novos.items():
    m = Chem.MolFromSmiles(smi)
    if m is None:
        print(f"SMILES inválida: {nome}")
        continue
    x = np.array([[f(m) for f in DESCRITORES.values()]])
    p = modelo_final.predict_proba(x)[0, 1]
    linhas.append({
        "composto": nome,
        "P(atravessa)": round(p, 3),
        "previsão": "atravessa" if p >= 0.5 else "não atravessa",
        "MW": round(Descriptors.MolWt(m), 1),
        "cLogP": round(Crippen.MolLogP(m), 2),
        "TPSA": round(rdmd.CalcTPSA(m), 1),
    })

pd.DataFrame(linhas)
````

Curcumina, resveratrol e quercetina são polifenóis de produtos naturais, com literatura enorme sobre efeitos neuroprotetores. Olhe as probabilidades que o modelo deu e o TPSA de cada um. A quercetina, com TPSA acima de 130, é justamente o caso clássico: atividade excelente in vitro, biodisponibilidade cerebral péssima in vivo.

Isso não invalida o interesse nesses compostos — é o motivo pelo qual tanta gente trabalha em nanoencapsulamento e pró-fármacos deles.

## <font color='green'>**7.7 Domínio de aplicabilidade: quando não acreditar no modelo**</font>

Todo modelo QSAR só vale dentro do espaço químico em que foi treinado. Fora dele, o modelo continua devolvendo um número com toda a confiança do mundo, e o número não significa nada.

Uma checagem simples: calcule a similaridade entre o composto novo e o vizinho mais próximo do conjunto de treino. Se o Tanimoto máximo for baixo, desconfie.

````python
#@title <font color='green'> 7.7.1 — Domínio de aplicabilidade
# Domínio de aplicabilidade por similaridade ao vizinho mais próximo.

LIMIAR = 0.25   # abaixo disso, considero o composto fora do domínio

print(f"{'composto':24s} {'Tc máx':>8s}  {'vizinho mais próximo':24s} situação")
print("-" * 78)
for nome, smi in novos.items():
    m = Chem.MolFromSmiles(smi)
    if m is None:
        continue
    fp = gerador.GetFingerprint(m)
    sims = DataStructs.BulkTanimotoSimilarity(fp, fps)
    j = int(np.argmax(sims))
    dentro = "dentro do domínio" if sims[j] >= LIMIAR else "FORA — não confie"
    print(f"{nome:24s} {sims[j]:8.3f}  {quimioteca.nome[j]:24s} {dentro}")
````

Boa parte dos produtos naturais aparece fora do domínio. Faz sentido: a quimioteca (*toy-dataset*) de treino é de fármacos sintéticos e não contém polifenóis nem terpenos. O modelo está extrapolando.

Isso é uma limitação estrutural do QSAR, não um defeito deste exercício. E é um problema prático sério para quem trabalha com produtos naturais no Nordeste: os modelos publicados foram quase todos treinados em bibliotecas farmacêuticas sintéticas, e a química de plantas e microrganismos da caatinga simplesmente não está representada neles.

# <font color='green'> **8. Exercício 4 — Da molécula 2D à estrutura 3D**

Até aqui a molécula foi um grafo: átomos e ligações, sem geometria. Agora ela precisa ganhar coordenadas no espaço, porque é assim que ela existe e é assim que ela encontra um alvo.

Este é o último passo da Parte 01 e a ponte para a Parte 02. Geramos conformações, calculamos energia com mecânica molecular e olhamos a estrutura em 3D. Na Parte 02 essa estrutura entra no sítio ativo de uma proteína.

## <font color='green'>**8.1 Gerando conformações**</font>

Uma molécula flexível não tem *uma* estrutura 3D. Ela tem um conjunto de conformações acessíveis, com populações que seguem a distribuição de Boltzmann. Ibuprofeno, com quatro ligações rotacionáveis, já tem dezenas.

O RDKit usa o algoritmo **ETKDG** (*experimental-torsion knowledge distance geometry*). Ele parte da matriz de distâncias interatômicas possíveis, sorteia coordenadas compatíveis e depois corrige os ângulos de torção usando preferências extraídas de estruturas cristalográficas do CSD. A versão 3 é a atual e é bem melhor que as anteriores para anéis grandes e macrociclos.

Depois de gerar, otimizamos com um campo de força. Uso o **MMFF94**, que é razoável para moléculas orgânicas pequenas.

````python
#@title <font color='green'> 8.1.1 — Busca conformacional
# Busca conformacional para o ibuprofeno.

smi_ibu = "CC(C)Cc1ccc(cc1)C(C)C(=O)O"
mol_ibu = Chem.AddHs(Chem.MolFromSmiles(smi_ibu))

print(f"Ligações rotacionáveis: {rdmd.CalcNumRotatableBonds(mol_ibu)}")

params = AllChem.ETKDGv3()
params.randomSeed = 0xC0FFEE      # semente fixa: resultado reprodutível
params.pruneRmsThresh = 0.5       # descarta conformações a menos de 0,5 Å

t0 = time.time()
ids = AllChem.EmbedMultipleConfs(mol_ibu, numConfs=200, params=params)
print(f"{len(ids)} conformações distintas geradas em {time.time()-t0:.1f} s")

# Otimização MMFF94: devolve (código_de_convergência, energia) por conformação
resultado = AllChem.MMFFOptimizeMoleculeConfs(mol_ibu, maxIters=2000)
energias = np.array([e for _, e in resultado])
convergiu = np.array([c for c, _ in resultado]) == 0

print(f"Convergiram: {convergiu.sum()} de {len(ids)}")

E_rel = energias - energias.min()
print(f"\nJanela energética: {E_rel.max():.2f} kcal/mol")
print(f"Conformações dentro de 3 kcal/mol do mínimo: {(E_rel < 3).sum()}")
````

## <font color='green'>**8.2 Populações de Boltzmann**</font>

Ter a conformação de menor energia não basta. A molécula em solução, a 37 °C, visita várias, e a proporção entre elas segue a distribuição de Boltzmann:

$$p_i = \frac{e^{-E_i/k_BT}}{\sum_j e^{-E_j/k_BT}}$$

A conta abaixo mostra quantas conformações somam 90% da população. Esse número é o que decide quantas estruturas você precisa levar para o docking.

````python
#@title <font color='green'> 8.2.1 — Populações de Boltzmann
# Populações de Boltzmann a 310 K (temperatura corporal, não 298).

kT = 0.001987 * 310.15                    # R·T em kcal/mol
pesos = np.exp(-E_rel / kT)
pop = pesos / pesos.sum()

ordem = np.argsort(E_rel)
print("Conformações mais povoadas a 37 °C:\n")
print(f"{'conf':>6s} {'E rel (kcal/mol)':>18s} {'população':>12s}")
for i in ordem[:6]:
    print(f"{i:6d} {E_rel[i]:18.4f} {100*pop[i]:11.2f}%")

fig, axes = plt.subplots(1, 2, figsize=(11.5, 3.8))
axes[0].hist(E_rel, bins=25, color="#7fcdbb", edgecolor="k", linewidth=0.4)
axes[0].set_xlabel("energia relativa (kcal/mol)")
axes[0].set_ylabel("nº de conformações")
axes[0].set_title("Distribuição de energias MMFF94")

axes[1].scatter(E_rel[ordem], 100*np.cumsum(pop[ordem]), s=22, color="#2c7fb8")
axes[1].axhline(90, color="crimson", ls="--", lw=1.2)
axes[1].set_xlabel("energia relativa (kcal/mol)")
axes[1].set_ylabel("população acumulada (%)")
axes[1].set_title("Quantas conformações somam 90% da população")
plt.tight_layout(); plt.show()
````

## <font color='magenta'>**8.4 Alinhando confôrmeros por um núcleo comum (opcional)**</font>

Quando você gera muitas conformações e quer visualizar todas sobrepostas, o alinhamento importa. Alinhar pela molécula inteira produz uma nuvem confusa. Alinhar por um fragmento rígido, o núcleo (ou *core*), deixa visível o que de fato se move.

````python
#@title <font color='magenta'> 8.4.3 — Sobreposição sem núcleo de referência
def generate_and_visualize_conformations(smiles, num_confs):
    # Gerar a molécula a partir do SMILES
    mol = Chem.MolFromSmiles(smiles)

    # Adicionar hidrogênios explícitos à molécula
    mol = Chem.AddHs(mol)

    # Gerar conformações
    confs = AllChem.EmbedMultipleConfs(mol, numConfs=num_confs)

    # Otimizar conformações usando UFF (Universal Force Field)
    for conf in confs:
        AllChem.UFFOptimizeMolecule(mol, confId=conf)
        #mol_without_H = Chem.RemoveHs(mol)

    # Alinhar todas as conformações ao primeiro conformero
    ref_conf = mol.GetConformer(confs[0])
    for conf in confs:
        if conf != confs[0]:
            rdMolAlign.AlignMol(mol, mol, prbCid=conf, refCid=confs[0])
            mol_without_H = Chem.RemoveHs(mol)

    # Gerar a visualização 3D
    # Visualizar as conformações usando py3Dmol
    viewer = p3d.view(width=800, height=600)
    for conf in confs:
        mol_block = Chem.MolToMolBlock(mol_without_H, confId=conf)
        viewer.addModel(mol_block, 'mol')  # Adicionar modelo ao visualizador
        viewer.setStyle({'stick': {}})  # Estilo de visualização

    viewer.zoomTo()  # Ajustar o zoom para visualizar todas as conformações
    return viewer

# Exemplo de uso
smiles = 'CC1=CN=C(C(=C1OC)C)CS(=O)C2=NC3=C(N2)C=C(C=C3)OC'  # Omeprazole
num = 25
viewer = generate_and_visualize_conformations(smiles,num)
viewer.show()
````

````python
#@title <font color='magenta'> 8.4.6 — Sobreposição alinhada pelo núcleo
def generate_and_visualize_conformations_core(smiles, core_smiles, num_confs):
    # Gerar a molécula a partir do SMILES
    mol = Chem.MolFromSmiles(smiles)

    # Adicionar hidrogênios explícitos à molécula
    mol = Chem.AddHs(mol)

    # Gerar conformações
    confs = AllChem.EmbedMultipleConfs(mol, numConfs=num_confs)

    # Otimizar conformações usando UFF (Universal Force Field)
    for conf in confs:
        AllChem.UFFOptimizeMolecule(mol, confId=conf)

    # Gerar o "core" (subestrutura comum) a partir do SMILES
    core_mol = Chem.MolFromSmiles(core_smiles)

    # Alinhar todas as conformações ao "core"
    # Encontre o mapeamento de átomos entre o core e a molécula
    match = mol.GetSubstructMatch(core_mol)
    if not match:
        raise ValueError("O core não é uma subestrutura da molécula.")

    atom_map = [(i, i) for i in match]

    for conf in confs:
        if conf != confs[0]:
            rdMolAlign.AlignMol(mol, mol, prbCid=conf, refCid=confs[0], atomMap=atom_map)

    # Visualizar as conformações usando py3Dmol
    viewer = p3d.view(width=800, height=600)
    for conf in confs:
        mol_block = Chem.MolToMolBlock(mol, confId=conf)
        viewer.addModel(mol_block, 'mol')  # Adicionar modelo ao visualizador
        viewer.setStyle({'stick': {}})  # Estilo de visualização

    viewer.zoomTo()  # Ajustar o zoom para visualizar todas as conformações
    return viewer

# Exemplo de uso
smiles = 'CC1=CN=C(C(=C1OC)C)CS(=O)C2=NC3=C(N2)C=C(C=C3)OC'  # Omeprazole
core_smiles = 'C2=NC3=C(N2)C=C(C=C3)OC'  # Core da molécula
num = 25
viewer = generate_and_visualize_conformations_core(smiles, core_smiles, num)
viewer.show()
````

- [Drug Design Org](https://www.drugdesign.org/) — É um portal com um extenso conteúdo didático. Tipo ChemLibreTexts.
- [Cheminformatics](https://chem.libretexts.org/Courses/Intercollegiate_Courses/Cheminformatics) — Curso online de quimioinformática acessível no portal [ChemLibreTexts](https://chem.libretexts.org/).



**Bancos de dados que você vai usar**

- [PubChem](https://pubchem.ncbi.nlm.nih.gov/) — mais de cem milhões de compostos, estruturas e bioensaios.
- [ChEMBL](https://www.ebi.ac.uk/chembl/) — dados de atividade biológica curados a partir da literatura.
- [Protein Data Bank](https://www.rcsb.org/) — estruturas experimentais de macromoléculas.
- [DrugBank](https://go.drugbank.com/) — fármacos aprovados, alvos e farmacocinética.
- [ZINC](https://zinc.docking.org/) — compostos comercialmente disponíveis, prontos para triagem virtual.
- [COCONUT](https://coconut.naturalproducts.net/) — produtos naturais, útil para quem trabalha com biodiversidade.

**Conjuntos de dados para modelos ADMET de verdade**

- [MoleculeNet](https://moleculenet.org/) — inclui o BBBP usado como referência no exercício 3.
- [Therapeutics Data Commons](https://tdcommons.ai/) — conjuntos padronizados de ADMET com divisões de treino e teste já definidas.
- B3DB — conjunto curado de permeabilidade da barreira hematoencefálica com valores de logBB.
